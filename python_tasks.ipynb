```
'''Use Case 2: Salary Processing System (HR / Payroll Automation)
Demonstrates input/output functions, default arguments, arbitrary args, returning values, and composing multiple functions.
What Problem It Solves:
Companies calculate employee salary differently depending on bonus percentage, incentives, PF, tax, etc.
How Function-Based Programming Helps:
We create multiple small reusable functions (bonus calc, tax calc, net salary calc)
Then compose them together.'''
```

```
basic_salary=float(input("Enter basic salary: "))
incentive1=float(input("Enter incentives: "))
incentive2=float(input("Enter incentives: "))
bonus_percent=float(input("Enter bonus percent: "))
def cal_incentives(*incentives):
    total_incentives=sum(incentives)
    return  total_incentives
def cal_pf(basic_salary,pf_percent=12):
    pf_amount=basic_salary*pf_percent/100
    return pf_amount
def cal_bonus(basic_salary,bonus_percent=10):
    bonus_amount=basic_salary*bonus_percent/100
    return bonus_amount

def cal_tax(gross_salary,tax_percent=5):
    tax_amount=gross_salary*tax_percent/100
    return tax_amount

def gross_salary(basic_salary,bonus_percent=10,*incentives):
    total_incentives=cal_incentives(*incentives)
    print(f"Total Incentives: {total_incentives}")
    total_bonus=cal_bonus(basic_salary,bonus_percent)
    print(f"Total Bonus: {total_bonus}")
    gross_salary=basic_salary+total_incentives+total_bonus
    print(f"gross_salary:{gross_salary}")
    pf_amount=cal_pf(basic_salary)
    print(f"pf_amount:{pf_amount}")
    tax_amount=cal_tax(gross_salary)
    print(f"tax_amount:{tax_amount}")
    total_salary_after_deduction=gross_salary-tax_amount-pf_amount
    print(f"total_salary_after_deduction:{total_salary_after_deduction}")
    return total_salary_after_deduction

print(f"Basic Salary Details: {basic_salary}")
print(f"Incentive1: {incentive1}")
print(f"Incentive2: {incentive2}")
print(f"Bonus: {bonus_percent}")

total_salary_after_deduction= gross_salary(basic_salary,bonus_percent,incentive1,incentive2)
```

```
'''Usecase3: Number Utility Tool
Create functions:
is_even(num) → returns True/False
find_max(*numbers) → returns highest number from arguments
perform_operation(num, operation) →
If operation = "square" → return num*num
If operation = "cube" → return numnumnum
If unknown → return "Invalid operation"


Test the functions with at least 5 numbers.'''
```

```
def is_even(num):
    if num % 2 == 0:
        return True
    else:
        return False
print(is_even(30))
def find_max(*numbers):
    max_num=max(numbers)
    return max_num
print(find_max(1,2,4,10,90))
def perform_operation(num,operation="addition"):
    if operation=="square":
        return num*num
    elif operation=="cube":
        return num*num*num
    else:
        return "Invalid Operation"

print(perform_operation(5,"cube"))

```
